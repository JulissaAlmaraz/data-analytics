# Exercise 2.A Using built-in functions
## Lab 2. Parsing strings - part codes

```python
def get_supplier_code(supplier):
    name = supplier.split(':')[0]
    return name
#here i am splitting whatever the argument may be in the function by the semicolon, then it turns into a list and i choosing the first [0] item in that list, and returning it.
```

```python
get_supplier_code('ACME:123-L')
```

```python
def get_product_num(prodnum):
  start = prodnum.find(':')
  end = prodnum.find('-')
  num = prodnum[start+1:end]
  return num
#Here i am creating different variables and 'finding' the index at which specifc characters are located. then assigning it to variable num, i am using the prodnum parameter and saying for whatever argument there is, i want to slice at the start and end. i added +1 since we want to start after the :
```

```python
get_product_num('ACME:123-L')
```

```python
def get_size(size):
  getsize = size.split('-')[-1]
  return getsize
get_size('ACME:123-L')
#Here it is similar to the the first one but i am accessing the LAST item.
```

```python
acme = 'ACME:123-L'
di = 'DI:123-M'
ace = 'ACE:1-12'
```

```python
supplier = get_supplier_code(acme)
part_num = get_product_num(acme)
size = get_size(acme)
print(f'For {acme} the supplier is {supplier}. Part number {part_num} is available in size {size}')
```

```python
supplier = get_supplier_code(di)
part_num = get_product_num(di)
size = get_size(di)
print(f'For {di} the supplier is {supplier}. Part number {part_num} is available in size {size}')
```

```python
supplier = get_supplier_code(ace)
part_num = get_product_num(ace)
size = get_size(ace)
print(f'For {ace} the supplier is {supplier}. Part number {part_num} is available in size {size}')
```
