## Lab 3. Employee Data
This lab exercise makes updates to information extracted from an employee database.

```python
hr_list = [
    ('0123', 'HR', 'Rebecca Yang', '69000'),
    ('0121', 'IT', 'Mark Blick', '67000'),
    ('0068', 'IT', 'Kahna Larsen', '112000'),
    ('0234', 'OPS', 'Jim Smith', '54000')
]
```

```python
for i, employee in enumerate(hr_list):
  if employee[2] == 'Mark Blick':
    hr_list[i] = (employee [0], employee[1], 'Mark Blick-Hawley', employee[3])
    print(hr_list)
```

```python
for i, employee in enumerate(hr_list):
  if employee[2] == 'Jim Smith':
    hr_list[i] = (employee[0], 'CS', employee[2], '60000')
print(hr_list)
```

```python
for employee in hr_list:
  emp_id = employee[0]
  dept_code = employee[1]
  name = employee[2]
  salary = int(employee[3])
  print(f'{emp_id} | {dept_code} | {name} | ${salary:,}')
```
