#Lab 2 of Exercise 2.B, working with datetime functions.

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
print(today)
```

```python
today_format = today.strftime('%A, %B %d, %Y')
print(today_format)
```

# My Birthday

```python
my_birhday = datetime.date(1999, 5, 4)
print(my_birhday)
```

```python
my_birthday_format = my_birhday.strftime('%m/%d/%Y')
print(my_birthday_format)
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days=90)
print(ninety_d)
```

```python
ninety_d_format = ninety_d.strftime('%B %d, %Y')
print(f'90 days from today is {ninety_d_format}')
```

# Time for Dinner

```python
dinner_time = datetime.time(19, 30)
print(dinner_time)
```

```python
dinner_time_format = dinner_time.strftime('%I:%M %p')
print(f'Lets meet for dinner at {dinner_time_format}')
```
