# Lab 1 of Exercise 2.C, working with defining custom functions

```python
def favorite_things(name, movie, music):
  return name, movie, music
print(favorite_things('Julissa', '28 Years Later', 'Experience'))
```

```python
def why_im_here(purpose):
  return purpose
print(why_im_here('I joined to gain experience and learn'))
```

```python
def favorite_place(place, why):
  return f'One of my favorite places is {place} because {why}'
print(favorite_place('Chicago', 'I grew up here'))
```
