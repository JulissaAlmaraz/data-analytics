# This exercise demonstrates the rates of taxes for social security, medicare, and federal tax.

```python
def get_soc_sec_tax(gross_pay):
  social_sec_tax = gross_pay * 0.062
  return social_sec_tax
```

```python
def get_medicare_tax(gross_pay):
  med_tax = gross_pay * 0.0145
  return med_tax
```

```python
def get_federal_tax(gross_pay, withhold_code):
  tax_rate = 0
  if withhold_code == 0:
    tax_rate = .23
  elif withhold_code == 1:
    tax_rate = .21
  elif withhold_code == 2:
    tax_rate = 0.195
  elif withhold_code == 3:
    tax_rate = 0.185
  elif withhold_code == 4:
    tax_rate = 0.18
  elif withhold_code > 4:
    tax_rate = 0.18 - ((withhold_code - 4) * 0.005)
  tax = gross_pay * tax_rate
  return tax
```

```python
get_soc_sec_tax(750)
```

```python
get_medicare_tax(750)
```

```python
get_federal_tax(750, 0)
```

```python
get_federal_tax(1550, 2)
```

```python
get_federal_tax(1100, 5)
```
