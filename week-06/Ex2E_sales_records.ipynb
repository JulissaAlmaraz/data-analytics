This notebook analyzes a small sales dataset using Python functions.
Each function takes the full dataset as its first argument.

```python
sales_data = [
   ('Laptop', 'East', 1200.00),
   ('Phone', 'West', 850.00),
   ('Tablet', 'East', 400.00),
   ('Monitor', 'South', 650.00),
   ('Keyboard', 'West', 75.00),
   ('Laptop', 'South', 1150.00),
   ('Phone', 'East', 900.00),
   ('Tablet', 'West', 425.00),
   ('Monitor', 'East', 680.00),
   ('Keyboard', 'South', 80.00),
   ('Laptop', 'West', 1300.00),
   ('Phone', 'West', 820.00),
]
```

## Lab 1. Total Sales by Region

```python
def regional_total(data, region):
  return sum(amt for _, r, amt in data if r == region)
```

```python
print(regional_total(sales_data, 'East'))
print(regional_total(sales_data, 'West'))
print(regional_total(sales_data, 'South'))
```

# Lab 2. Count of sales by region

```python
def region_count(data, region):
  return sum(1 for _, r, amt in data if r == region)
```

```python
print(region_count(sales_data, 'East'))
```

# Lab 3. Average sale amount by region

```python
def region_average(data, region):
  count = region_count(data, region)
  if count == 0:
    return 0
  else:
    total = regional_total(data, region)
    average = total / count
    return average
```

```python
print(region_average(sales_data, 'North'))
print(region_average(sales_data, 'East'))
print(region_average(sales_data, 'West'))
```

# Lab 4. Filter by sale amount

```python
def sales_above(data, threshold):
  above = []
  for record in data:
    if record[2] > threshold:
      above.append(record)
  return above
```

```python
print(sales_above(sales_data, 1000))
```

```python
def sales_below(data, threshold):
  below = []
  for record in data:
    if record[2] < threshold:
      below.append(record)
  return below
```

```python
print(sales_below(sales_data, 500))
```

# Lab 5. Top and bottom sale

```python
def top_sale(data):
  return max(data, key=lambda s: s[2])
```

```python
def bottom_sale(data):
    return min(data, key=lambda s: s[2])
```

```python
print(top_sale(sales_data))
print(bottom_sale(sales_data))
```

The top sale was a laptop in the West region for $1300.

The bottom sale was a keyboard in the West region for $75.00.

# Lab 6. Apply a discount

```python
def apply_discount(data, discount_pct):
 return list(map(lambda s: (s[0], s[1], s[2] * (1 - discount_pct)), data))
```

```python
print(sales_data)
```

```python
apply_discount(sales_data, .15)
```

# Lab 7. Unique regions

```python
def unique_regions(data):
  return sorted(set(region for _, region, amt in data ))
```

```python
print(unique_regions(sales_data))
```

This function could be useful in a larger analysis program because it helps organize and narrow down data into specific categories.

# Lab 8. Region summary

```python
def region_summary(data, region):
  total = regional_total(data, region)
  count = region_count(data, region)
  average = region_average(data, region)
  region_data = [record for record in data if record[1] == region]
  top = top_sale(region_data)

  print(f'== {region} ==\nRecords: {count}\nTotal Sales: ${total:.2f}\nAverage Sales: ${average:.2f}\nTop Sale: {top[0]} - ${top[2]:.2f}')
```

```python
region_summary(sales_data, 'East')
```

# Lab 9. Full Summary

```python
for region in unique_regions(sales_data):
 summary = region_summary(sales_data, region)
 print(summary)

```
