## Lab 1. Working with names - parsing strings


```python
full_name = "John Smith"
```

```python
def parse_and_display(name):
  for i in range(len(name)):
    if name[i] == ' ':
      first = name[:i]
      last = name[i+1:]
      print(f'Full name: {name}')
      print(f'First name: {first}')
      print(f'Last name: {last}')
```

```python
parse_and_display(full_name)
```

```python
test_list = ['John Smith', 'Jane Doe', 'Sponge Bob']
```

```python
parse_and_display(test_list[0])
parse_and_display(test_list[1])
parse_and_display(test_list[2])
```

```python
parse_and_display('Julissa Almaraz Flores')
```

```python
def parse_and_display2(name):
  if name.count(' ') <= 1:
    for i in range(len(name)):
      if name[i] == ' ':
        first = name[:i]
        last = name[i+1:]
  elif name.count(' ') > 1:
    first, last = name.split(" ",1)
  print(f'Full name: {name}')
  print(f'First name: {first}')
  print(f'Last name: {last}')
```

```python
parse_and_display2('Julissa Almaraz Flores')
parse_and_display2('John Smith')
```
