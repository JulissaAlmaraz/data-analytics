```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13], [14, 15, 16]])
print(arrA)
```

```python
for num in arrA:
  print(num * 2)
```

```python
for num1 in arrA:
  for num2 in num1:
    print(num2 * 2)
```

For the first cell, it prints an entire row. The second one outputs the numbers one at a time

```python
for num1 in arrA:
  for num2 in num1:
      loop_result = num1 * 2
      print(loop_result)
#The output is now repeated rows when we multiply with num1. Because num1 is staying the same throughout, so we are looping through num2 but num1 is still staying the same the whole time
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA), axis = 0)
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA), axis=1)
print(joined_array_2)
```

```python
stacked_array_1 = np.stack((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
#stacked_array_2 = np.stack((joined_array_1, joined_array_2))
#I have a value error here because the arrays must have the same shape, and the arrays above do not. the dimensions have to match correctly to be able to stack, one array has 3 columns and the other has 6 so they will not stack
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
#another_2 = np.concatenate((joined_array_1, arrA), axis=1)
#print(another_2)
#This throws a ValueError because there is not enough rows in arrA to match the rows in joined_array_1
```

```python
a_hstack = np.hstack((arrA, arrA))
print(a_hstack)
#This adds columns but the number rows have to match

a_vstack = np.vstack((arrA, arrA))
print(a_vstack)
#This adds rows but the columnds must match

a_dstack = np.dstack((arrA, arrA))
print(a_dstack)
#This stacks along height and depth
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA, 'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7,6], [13,5], [7,3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)

print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
print(arrB)
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
