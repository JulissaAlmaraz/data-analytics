## Generating Random Numbers

## Generating random Choices from an Array

```python
students = np.array(['Alexander', 'Abraham', 'Max', 'Angel', 'Vanessa', 'Elizabeth', 'Gerardo', 'Blanca', 'Antonio', 'Bella', 'Jared', 'Jonathan'])
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size=(3,4))
print(breakout_groups)
#This would not work for real breakout room groups because we get repeated names, so it isn't realistic.
```

```python
call_on_weighted = np.random.choice(
    students,
    size = 5,
    p=[ 0.3,
        0.2,
        0.1,
        0.1,
        0.05,
        0.05,
        0.05,
        0.05,
        0.03,
        0.03,
        0.04,
        0.0 ]

    )
print(call_on_weighted)
# I assigned 30% probability to index 0 and 0% probability to the last index
```

## Generating a normal distribution

```python
bell_curve = np.random.normal(loc=10, scale=5, size=(4,5))
print(bell_curve)
```
