## Permutations

```python
print(r_4)
print(np.random.shuffle(r_4))
#The original array shuffled after adding the function, the second line returned 'None'
```

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
#Here, both arrays are being printed, so permutaion is not overriding the original array. We would use permuation instead of shuffle if we plan to keeo our original array
```

## Grocery Problem

```python
bag1 = np.array([2, 3, 4, 0])
bag2 = np.array([4, 0, 1, 2])
print(bag1)
print(bag2)
```

```python
print(f'This first bag contains {np.sum(bag1)} items, and the second bag contains {np.sum(bag2)} items')
```

```python
total = bag1 + bag2
print(f'There are {total[0]} vegetables, {total[1]} fruits, {total[2]} boxes of snacks, and {total[3]} beverages in the two grocery bags.')
```

## Teamwork

```python
Ali = np.array([5, 4, 6, 4, 2])
```

```python
Ben = np.array([0, 5, 5, 3, 4])
```

```python
Carla = np.array([3, 6, 0, 3, 4])
```

```python
all_knocks = Ali + Ben + Carla
print(all_knocks)
```

```python
ali_week = np.sum(Ali)
print(ali_week)
```

```python
ben_week = np.sum(Ben)
print(ben_week)
```

```python
carla_week = np.sum(Carla)
print(carla_week)
```

```python
group_week = np.sum(all_knocks)
print(group_week)
```

```python
rolling_total = np.cumsum(all_knocks)
print(rolling_total)
```

```python
estimated_votes = rolling_total / 2
print(estimated_votes)
```
